```python
df['gender'] = df['gender'].astype(str).str.strip().str.lower()

gender_map_lower = {
    'Male': 'Male', 'M': 'Male', 'male': 'Male', '1': 'Male',
    'Female': 'Female', 'F': 'Female', 'female': 'Female', '0': 'Female'
}

df['gender'] = df['gender'].map(gender_map_lower).fillna('Unknown')
```

```python
df['gender'].value_counts(dropna=False)
```

```python
df['gender'].unique()
```

```python
df['follow_up_required'] = df['follow_up_required'].astype(str).str.strip().str.lower()

followup_map_lower = {
    'yes': 'Yes', 'y': 'Yes', '1': 'Yes',
    'no': 'No', 'n': 'No', '0': 'No'
}
df['follow_up_required'] = df['follow_up_required'].map(followup_map_lower).fillna('No')
```

```python
df.head(10)
```

```python
gender_mode = df['gender'].mode()[0] if not df['gender'].mode().empty else 'Unknown'
followup_mode = df['follow_up_required'].mode()[0] if not df['follow_up_required'].mode().empty else 'No'
```

```python
print(gender_mode)
print(followup_mode)
```

```python
df['gender'].fillna(gender_mode,inplace=True)

df['follow_up_required'].fillna(followup_mode,inplace=True)
```

```python
print(df['gender'].value_counts(dropna=False ))
print(df['follow_up_required'].value_counts(dropna=False))
```

```python
df['billing_amount'] = df['billing_amount'].astype(str)
```

```python
df['billing_amount'] = df['billing_amount'].str.replace(r'[^0-9.]', '', regex=True)
```

```python
print("\n清洗掉字符后的样例（前10个）：")
print(df['billing_amount'].head(10))
```

```python
df['billing_amount'] = pd.to_numeric(df['billing_amount'], errors='coerce')
```

```python
print(f"\n转换后，计费金额列的空值数量: {df['billing_amount'].isna().sum()}")
```

```python
median_billing = df['billing_amount'].median()
print(f"计费金额的中位数是: {median_billing}")
```

```python
df['billing_amount']=df['billing_amount'].fillna(median_billing)
```

```python
print(f"填充后，计费金额列的空值数量: {df['billing_amount'].isna().sum()}")
```
